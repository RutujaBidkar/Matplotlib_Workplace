```python
import matplotlib.pyplot as plt
```

```python
marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70,
         72, 75, 78, 80, 82, 85, 88, 90, 95, 150]
plt.violinplot(marks)
plt.title("Marks Distribution")
plt.grid(axis="y")
plt.show()
```

```python
class_A = [45, 50, 55, 60, 65, 70, 72, 75, 80, 85]
class_B = [50, 55, 60, 65, 70, 75, 78, 82, 88, 92]
plt.violinplot([class_A, class_B])
plt.xticks([1, 2], ["Class A", "Class B"])
plt.title("Class Marks Distribution")
plt.xlabel("Class A and Class B")
plt.ylabel("Marks")
plt.grid(axis="y")
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]

sales = [100, 150, 120, 180]
expenses = [80, 100, 90, 130]

fig, ax = plt.subplots(1, 2)

ax[0].plot(months, sales)
ax[0].set_title("Sales")

ax[1].bar(months, expenses)
ax[1].set_title("Expenses")

plt.show()
```
