```python
import matplotlib.pyplot as plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [2000, 3000, 2500, 4000, 5000]
plt.plot(months, sales, marker="o", markersize=10)

plt.show()
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
orders = [20, 35, 30, 45, 50]
plt.plot(days,orders,marker="o", markersize =12)
plt.show()
```

```python


days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
orders = [20, 35, 30, 45, 50]

plt.plot(
    days,
    orders,
    marker="o",
    markersize=10,
    markeredgecolor="black"
)

plt.show()
```

```python
products = ["A", "B", "C", "D"]
sales = [100, 150, 120, 180]
plt.plot(
    products,
    sales,
    marker="o",
    markersize=10,
    markeredgecolor="black"
)

plt.show()
```

```python
products = ["A", "B", "C", "D"]
sales = [100, 150, 120, 180]

plt.plot(
    products,
    sales,
    marker="o",
    markersize=10,
    markeredgecolor="black",
    markerfacecolor="yellow"
)

plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr"]
profit = [2000, 3000, 2500, 4000]

plt.plot(
    months,
    profit,
    marker="o",
    linewidth=3
)

plt.show()
```
