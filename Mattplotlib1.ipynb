```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.plot(x,y)
plt.show()
```

```python
month= ["Jan","Feb","Mar","Apr","May","Jun"]
sales= [8000,7000,9222,8523,8888,9633]
plt.plot(month,sales,marker = "o")
plt.grid()
plt.xlabel("Month")
plt.ylabel("sales")
plt.title("Monthly Sales Report")
plt.show()
```

```python
Hour = [1,2,3,4,5]
Marks = [40,50,60,70,80]
plt.scatter(Hour,Marks)
plt.xlabel("Study Hour")
plt.ylabel("Marks")
plt.title("Study Hour vs Marks")
plt.show()
```
