```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]

sales = [100, 150, 120, 180]
expenses = [80, 100, 90, 130]

fig, ax = plt.subplots(2, 2, figsize=(10, 7))

# 1. Line Chart
ax[0, 0].plot(months, sales, marker="o")
ax[0, 0].set_title("Monthly Sales")
ax[0, 0].set_xlabel("Month")
ax[0, 0].set_ylabel("Sales")

# 2. Bar Chart
ax[0, 1].bar(months, expenses)
ax[0, 1].set_title("Monthly Expenses")
ax[0, 1].set_xlabel("Month")
ax[0, 1].set_ylabel("Expenses")

# 3. Scatter Plot
ax[1, 0].scatter(months, sales, s=80)
ax[1, 0].set_title("Sales Scatter Plot")
ax[1, 0].set_xlabel("Month")
ax[1, 0].set_ylabel("Sales")

# 4. Bar Chart
ax[1, 1].bar(months, sales)
ax[1, 1].set_title("Sales Bar Chart")
ax[1, 1].set_xlabel("Month")
ax[1, 1].set_ylabel("Sales")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]

sales = [100, 150, 120, 180]
expenses = [80, 100, 90, 130]

fig, ax = plt.subplots(2, 1, sharex=True, figsize=(8, 7))

# Top graph — Sales
ax[0].plot(months, sales, marker="o")
ax[0].set_title("Monthly Sales")
ax[0].set_ylabel("Sales")
ax[0].grid()

# Bottom graph — Expenses
ax[1].bar(months, expenses)
ax[1].set_title("Monthly Expenses")
ax[1].set_xlabel("Month")
ax[1].set_ylabel("Expenses")
ax[1].grid()

plt.tight_layout()
plt.show()
```
