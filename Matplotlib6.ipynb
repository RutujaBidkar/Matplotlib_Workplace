```python
import matplotlib.pyplot as plt
```

```python
products = ["Laptop", "Phone", "Tablet", "Watch"]

sales_2025 = [120, 200, 150, 100]
sales_2026 = [150, 220, 180, 130]

fig, ax = plt.subplots(2, 1, sharey=True, figsize=(8, 7))

# Graph 1 — Sales 2025
ax[0].bar(products, sales_2025)
ax[0].set_title("Sales 2025")
ax[0].set_ylabel("Sales")
ax[0].grid(axis="y")

# Graph 2 — Sales 2026
ax[1].bar(products, sales_2026)
ax[1].set_title("Sales 2026")
ax[1].set_xlabel("Products")
ax[1].set_ylabel("Sales")
ax[1].grid(axis="y")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [100, 150, 120, 180, 160, 200]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

# Y-axis limit
plt.ylim(0, 250)

# Rotate X-axis labels
plt.xticks(rotation=45)

plt.grid()

plt.show()
```
