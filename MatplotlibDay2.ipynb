```python
import matplotlib.pyplot as plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [8000, 7000, 9222, 8523, 8888, 9633]
plt.plot(
    months,
    sales,
    marker="o",
    color="red",
    linestyle="--",
    linewidth=2,
    markersize=8
)

plt.plot(months,sales)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

```python

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales_2025 = [8000, 7000, 9222, 8523, 8888, 9633]
sales_2026 = [9000, 8500, 9500, 9200, 10000, 11000]

plt.plot(
    months,
    sales_2025,
    marker="o",
    label="2025"
)

plt.plot(
    months,
    sales_2026,
    marker="*",
    label="2026"
)

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.grid()

plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales_2025 = [8000, 7000, 9222, 8523, 8888, 9633]
sales_2026 = [9000, 8500, 9500, 9200, 10000, 11000]

plt.plot(
    months,
    sales_2025,
    marker="o",
    linestyle="--",
    color="blue",
    linewidth=2,
    label="2025"
)

plt.plot(
    months,
    sales_2026,
    marker="*",
    linestyle=":",
    color="green",
    linewidth=2,
    label="2026"
)

plt.title("Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.legend()

plt.show()
```

```python
products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [120, 200, 150, 100]
plt.bar(products, sales,color ="blue")
plt.title("products sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.grid(axis="y")
plt.legend()
plt.show()
```

```python

products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [120, 200, 150, 100]

plt.bar(products, sales, color="green", width=0.6)

for i in range(len(products)):
    plt.text(i, sales[i], str(sales[i]))

plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.grid(axis="y")

plt.show()
```

```python
products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [120, 200, 150, 100]
plt.barh(products,sales,color="purple")
plt.title("products sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.grid(axis="x")
```

```python
hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [45, 50, 55, 60, 65, 72, 78, 85]
plt.scatter(
    hours,
    marks,
    color="red",
    marker="o",
    s=80
)
plt.title("study Hour vs Marks")
plt.xlabel("study hour")
plt.ylabel("Marks")
plt.grid()
```

```python
hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [45, 50, 55, 60, 65, 72, 78, 85]
plt.scatter(
    hours,
    marks,
    color="red",
    marker="o",
    alpha=0.5,
    s=80
)
plt.title("study Hour vs Marks")
plt.xlabel("study hour")
plt.ylabel("Marks")
plt.grid()
```

```python
marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70,
         72, 75, 78, 80, 82, 85, 88, 90, 92, 95]
plt.hist(
    marks,
    bins =5,
    color= "skyblue",
    edgecolor="black",
)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.grid()
```

```python
marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70,
         72, 75, 78, 80, 82, 85, 88, 90, 92, 95]
plt.hist(
    marks,
    bins=6,
    density=True,
    color ="skyblue",
    edgecolor="black",
)
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Density")
plt.grid()
```
