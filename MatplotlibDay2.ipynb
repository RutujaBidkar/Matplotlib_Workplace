```python
import matplotlib.pyplot as plt
```

```python
temperature = [20,25,30,35,40,45]
icecream_sales = [100,150,220,300,400,500]

plt.scatter(temperature, icecream_sales)

plt.xlabel("Temperature")
plt.ylabel("Ice Cream Sales")
plt.title("Temperature vs Ice Cream Sales")

plt.show()
```

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [500, 800, 300, 600]

plt.bar(products, sales)
plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Product Sales")

plt.show()
```

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [500, 800, 300, 600]

plt.barh(products, sales)

plt.xlabel("Sales")
plt.ylabel("Products")
plt.title("Product Sales")

plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 60, 62, 65, 68, 70, 72,
         75, 78, 80, 82, 85, 88, 90, 92, 95]

plt.hist(marks)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Marks Distribution")

plt.show()
```

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [500, 800, 300, 600]

plt.pie(sales, labels=products)

plt.title("Product Sales Distribution")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [4000, 5000, 4500, 7000]

plt.plot(months, sales, color="red")

plt.show()
```
