```python
import matplotlib.pyplot as plt
```

```python
marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70,
         72, 75, 78, 80, 82, 85, 88, 90, 92, 95]

plt.hist(
    marks,
    bins=6,
    density=True,
    color="orange",
    edgecolor="black"
)

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Density")
plt.grid()

plt.show()
```

```python
products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [120, 200, 150, 100]

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Product Sales Share")

plt.show()
```

```python
marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70,
         72, 75, 78, 80, 82, 85, 88, 90, 92, 95]

plt.boxplot(marks)

plt.title("Marks Distribution")
plt.ylabel("Marks")
plt.grid(axis="y")

plt.show()
```

```python
marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70,
         72, 75, 78, 80, 82, 85, 88, 90, 95, 150]

plt.boxplot(marks)

plt.title("Marks With Outlier")
plt.ylabel("Marks")
plt.grid(axis="y")

plt.show()
```
